# Data Auditing & Cleaning Phase

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("./data/Leads.csv")
df.head(10)
```

**Replacing null placeholders with null**

```python
df = df.replace("Select",np.nan)
df.head(10)
```

## Dataset Summary

The results below suggest that :-
- Columns with missing values greater than 50% should be removed as filling them would basically be generating new data.
- Columns with previous index and score whould also be removed to avoid data leakage.
- Columns with high cardinality such as IDs should be removed or adjusted for one hot encoding.
- Missing values should be replaced with appropriate placeholder values.

**Checking the rate of missing values to understand which columns to remove and which columns to modify**

```python
def missing_values(df):
    missing_rate = df.isnull().mean().round(4).sort_values(ascending=False)*100
    return missing_rate

missing_values(df)
```

**Checking unique values to remove/adjust high cardinality columns for one hot encoding**

```python
def check_unique_values(df):
    unique_values = df.select_dtypes("object").nunique().sort_values(ascending=False)
    return unique_values

check_unique_values(df)
```

**Checking if the target column is well balanced for training**

```python
def target_column_distribution(df):
    target_normalization = df["Converted"].value_counts(normalize=True)
    return target_normalization

target_column_distribution(df)
```

## Data Cleaning

**Dropping the straighforward unnecessary columns (Include IDs, Unnecessary Info, Redundent Info)**

```python
drop_list = ["Lead Quality","Lead Profile","How did you hear about X Education","Asymmetrique Activity Index", "Last Notable Activity",
             "Asymmetrique Profile Index","Asymmetrique Activity Score","Asymmetrique Profile Score","Prospect ID"]
df.drop(drop_list,axis=1,inplace=True)
df.head(5)
```

**Changing the null values for categorical data to "Unknown" and the median value for numeric data.**

```python
def data_cleaning(df):    
    object_cols = df.select_dtypes("object")
    df[object_cols.columns] = object_cols.fillna("Unknown")

    numeric_cols = df.select_dtypes("number")
    df[numeric_cols.columns] = numeric_cols.fillna(numeric_cols.median())
    return df
df = data_cleaning(df)
```

Results below show that
- "Country" column is mostly one single category making it ideal for a binary column.
- The other two columns are well distributed among the top 5 categories dominating most of the data indicating that all 5 categories should be preserved.

All other categories of each column will be categorized as "Other".

**Changing all other categories to "Other" for "Country" column except unknown values.**

```python
df["Country"] = list(map(lambda x: x if (x=="India" or x=="Unknown") else "Other",df["Country"]))
df["Country"].value_counts()
```

**Below we checked the porportion of data for top 5 categories for each high cardinality column to ensure which method to approach to reduce cardinality.**

```python
high_cardinality = ["Country","Tags","Lead Source","Specialization","Last Activity"]
def card_check(df,card,top_n=5):
    for col in card:
        print(df[col].value_counts(normalize=True).head(top_n)*100)
        print("\n")
card_check(df,high_cardinality)
```

**Changing other high cardinality columns to 6 distinct categories only.**

```python
def reduce_cardinality(df,cols):
    for col in cols:
        other_cat = df[col].value_counts(ascending=True)[0:-5]
        df[col] = list(map(lambda x: "Other" if x in other_cat else x,df[col]))
    return df
df = reduce_cardinality(df,high_cardinality[1:])
card_check(df,high_cardinality,6)
```
