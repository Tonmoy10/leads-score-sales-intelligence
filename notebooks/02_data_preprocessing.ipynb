# Data Preprocessing & Clustering

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/interim/Cleaned_Leads.csv")
df.head(10)
```

**Dropping Target column for Unsupervised Learning to avoid Data Leakage**

```python
target_column = "Converted"
df.drop(columns=[target_column])
```

```python
df.shape
```

**Applying one hot encoding for categorical columns**

```python
df_encoded = pd.get_dummies(df,drop_first=True)
df_encoded.info     
```

**Standardizing the values as few fields have much larger scale of values compared to others**

```python
stdsclr = StandardScaler()
scaled_data = stdsclr.fit_transform(df_encoded)
df_processed = pd.DataFrame(data=scaled_data,columns=df_encoded.columns)
df_processed.describe().round(2)
```

```python
df_processed.to_csv("../data/processed/Processed_Leads.csv",index=False)
```
