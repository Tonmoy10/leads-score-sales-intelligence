# Data Preprocessing & Clustering

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("data/Cleaned_Leads.csv")
df.head(10)
```

**Dropping Target column for Unsupervised Learning to avoid Data Leakage**

```python
target_column = "Converted"
df.drop(columns=[target_column])
```

```python
df.select_dtypes("object").nunique().sort_values(ascending=False)
```

```python
df.shape
```

```python
df_encoded = pd.get_dummies(df,drop_first=True)
df_encoded.info     
```

```python
stdsclr = StandardScaler()
scaled_data = stdsclr.fit_transform(df_encoded)
df_new = pd.DataFrame(data=scaled_data,columns=df_encoded.columns)
df_new.describe().round(2)
```
